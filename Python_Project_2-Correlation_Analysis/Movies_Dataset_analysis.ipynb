```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv('movies.csv')
```

```python
df
```

### To see if there's any missing data


```python
for col in df.columns:
    pct_missing = np.mean(df[col].isnull())
    print(f'{col} - {pct_missing:.2f} %')
```

```python
df.dtypes
```

### Dropping NaN Values from Columns

```python
df =  df.dropna(subset = ['budget'])
df = df.dropna(subset = ['gross'])
df
```

### Changing DataTypes of Columns

```python
df['budget'] = df['budget'].astype('int64')
df['gross'] = df['gross'].astype('int64')
df
```

### Converting Released Column to DateTime Format

```python
# df['released'] = df['released'].str.split(' (', regex=False).str[0]
df['released'] = pd.to_datetime(df['released'], format='mixed', errors='coerce')
df['correctyear'] = df['released'].dt.year
df
```

### Sorting values of the 'Gross' Column in Descending order

```python
df.sort_values(by = ['gross'], inplace = True, ascending = False)
```

### Dropping any duplicate values from Company Column

```python
df['company'] = df['company'].drop_duplicates()
```

### Plotting a Scatter Plot Graph to see Correlation between Budget and Gross Revenue

```python
plt.figure(figsize = (6,5))
plt.scatter(x= df['budget'], y= df['gross'])
plt.xlabel('Budget (USD 100 Million)')
plt.ylabel('Gross Revenue (USD 100 Million)')
plt.title('Budget Vs. Gross Revenue')
```

```python
plt.figure(figsize = (8,7))
sns.regplot(data = df, x= 'budget', y = 'gross', scatter_kws= {'color': 'red'}, line_kws= {'color': 'blue'})

plt.xlabel('Budget (USD 100 Million)')
plt.ylabel('Gross Revenue (USD 100 Million)')
plt.title('Budget Vs. Gross Revenue')
```

### Finding Correlations between Every column that has Numeric Values

```python
df.corr(numeric_only= True)
```

### Plotting a HeatMap based on that Correlation

```python
correlation_matrix = df.corr(numeric_only= True)
sns.heatmap(data = correlation_matrix, annot= True)
```

```python
df.columns
```

### Filling Values of every non-numeric column 

```python
df_normalized = df

for col_name in df_normalized.columns:
    if(df_normalized[col_name].dtype == 'object'):
        df_normalized[col_name] = df_normalized[col_name].astype('category')
        df_normalized[col_name] = df_normalized[col_name].cat.codes

df_normalized.head(20)
```

### Plotting a HeatMap based on that Correlation between all the columns in the dataframe

```python
plt.figure(figsize = (13,6))

correlation_matrix = df_normalized.corr(numeric_only= True)
sns.heatmap(data = correlation_matrix, annot= True)
```

### Sorting Pairs of Columns based on their correlation between them

```python
df_pairs = df_normalized.corr().unstack()
sorted_pairs = df_pairs.sort_values()
sorted_pairs
```

### Finding every pair of column with correlation more than 0.5 

```python
high_corr = sorted_pairs[(sorted_pairs > 0.5)]
high_corr
```
